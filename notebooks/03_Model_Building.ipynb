```python
# Step 1: Import required libraries

import pandas as pd
import numpy as np

# Load the processed dataset
processed_path = r"C:\Users\akula\OneDrive\Desktop\DOCUMENTS\Seanergy_Preparation\Projects\Automated_Email_Classification\data\processed\spam_processed.csv"

df = pd.read_csv(processed_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
# Step 2: Check target labels

print("Target value counts:")
print(df["target"].value_counts())

print("\nTarget value types:")
print(df["target"].unique())
```

```python
# Step 3: Prepare input (X) and target (y)

# Input text
X = df["transformed_text"]

# Target label
y = df["target"].map({"ham": 0, "spam": 1})

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFirst 5 input texts:")
print(X.head())

print("\nFirst 5 target values:")
print(y.head())
```

```python
# Step 4: Split the dataset into training and testing sets

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining labels:")
print(y_train.value_counts())

print("\nTesting labels:")
print(y_test.value_counts())
```

```python
# Step 5: Convert text into numerical features using TF-IDF

from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95
)


# Learn vocabulary only from training data
X_train_tfidf = tfidf.fit_transform(X_train)

# Transform testing data using the same vocabulary
X_test_tfidf = tfidf.transform(X_test)

print("TF-IDF vectorization completed successfully!")

print("\nTraining TF-IDF shape:", X_train_tfidf.shape)
print("Testing TF-IDF shape:", X_test_tfidf.shape)
```

```python
# Step 6: Train Multinomial Naive Bayes

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Create the model
nb_model = MultinomialNB()

# Train the model
nb_model.fit(X_train_tfidf, y_train)

# Make predictions
y_pred_nb = nb_model.predict(X_test_tfidf)

# Calculate accuracy
nb_accuracy = accuracy_score(y_test, y_pred_nb)

print("Naive Bayes Model")
print("-----------------")
print("Accuracy:", nb_accuracy)

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_nb,
    target_names=["Ham", "Spam"]
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_nb))
```

```python
# Step 7: Train Logistic Regression

from sklearn.linear_model import LogisticRegression

# Create the model
lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train the model
lr_model.fit(X_train_tfidf, y_train)

# Make predictions
y_pred_lr = lr_model.predict(X_test_tfidf)

# Calculate accuracy
lr_accuracy = accuracy_score(y_test, y_pred_lr)

print("Logistic Regression Model")
print("-------------------------")
print("Accuracy:", lr_accuracy)

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_lr,
    target_names=["Ham", "Spam"]
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_lr))
```

```python
# Step 8: Train Support Vector Machine (SVM)

from sklearn.svm import LinearSVC

# Create the model
svm_model = LinearSVC(random_state=42)

# Train the model
svm_model.fit(X_train_tfidf, y_train)

# Make predictions
y_pred_svm = svm_model.predict(X_test_tfidf)

# Calculate accuracy
svm_accuracy = accuracy_score(y_test, y_pred_svm)

print("Support Vector Machine (SVM)")
print("----------------------------")
print("Accuracy:", svm_accuracy)

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_svm,
    target_names=["Ham", "Spam"]
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_svm))
```

```python
# Step 9: Compare all models

from sklearn.metrics import precision_score, recall_score, f1_score

model_comparison = pd.DataFrame({
    "Model": [
        "Multinomial Naive Bayes",
        "Logistic Regression",
        "Linear SVM"
    ],
    "Accuracy": [
        accuracy_score(y_test, y_pred_nb),
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_svm)
    ],
    "Spam Precision": [
        precision_score(y_test, y_pred_nb),
        precision_score(y_test, y_pred_lr),
        precision_score(y_test, y_pred_svm)
    ],
    "Spam Recall": [
        recall_score(y_test, y_pred_nb),
        recall_score(y_test, y_pred_lr),
        recall_score(y_test, y_pred_svm)
    ],
    "Spam F1-Score": [
        f1_score(y_test, y_pred_nb),
        f1_score(y_test, y_pred_lr),
        f1_score(y_test, y_pred_svm)
    ]
})

print(model_comparison.round(4).to_string(index=False))
```

```python
# Step 10: Save the trained model and TF-IDF vectorizer

import os
import joblib

# Create model directory
model_dir = r"C:\Users\akula\OneDrive\Desktop\DOCUMENTS\Seanergy_Preparation\Projects\Automated_Email_Classification\models"

os.makedirs(model_dir, exist_ok=True)

# Save the SVM model
svm_model_path = os.path.join(model_dir, "spam_classifier_svm.pkl")
joblib.dump(svm_model, svm_model_path)

# Save the TF-IDF vectorizer
tfidf_path = os.path.join(model_dir, "tfidf_vectorizer.pkl")
joblib.dump(tfidf, tfidf_path)

print("Model and vectorizer saved successfully!")
print("\nSVM model:")
print(svm_model_path)

print("\nTF-IDF vectorizer:")
print(tfidf_path)
```

```python
# Step 11: Load the saved model and TF-IDF vectorizer

import joblib

# Load SVM model
svm_model = joblib.load(svm_model_path)

# Load TF-IDF vectorizer
tfidf = joblib.load(tfidf_path)

print("SVM model loaded successfully!")
print("TF-IDF vectorizer loaded successfully!")
```

```python
# Step 12: Create a function to classify new emails

def predict_email(email_text):
    # Convert email text into TF-IDF features
    email_tfidf = tfidf.transform([email_text])
    
    # Predict the class
    prediction = svm_model.predict(email_tfidf)[0]
    
    return prediction
```

```python
# Step 13: Test with a spam email

test_email = "Congratulations! You have won a free prize. Click here to claim your reward."

prediction = predict_email(test_email)

print("Email:", test_email)
print("Prediction:", prediction)
```

```python
# Step 14: Test with a legitimate email

test_email = "Hi Priya, the project meeting is scheduled for tomorrow at 10 AM. Please bring the updated report."

prediction = predict_email(test_email)

print("Email:", test_email)
print("Prediction:", prediction)
```

```python
# Step 15: Convert prediction into readable label

def classify_email(email_text):
    # Convert text into TF-IDF features
    email_tfidf = tfidf.transform([email_text])
    
    # Predict
    prediction = svm_model.predict(email_tfidf)[0]
    
    # Convert numeric prediction to readable label
    if prediction == 1:
        return "Spam"
    else:
        return "Ham"
```
